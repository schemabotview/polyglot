## Logical Operators & Truthiness

### Logical operators

| Sub-feature | Java | Scala | Kotlin | JavaScript | TypeScript | Python |
|---|---|---|---|---|---|---|
| logical AND | `a && b` | `a && b` | `a && b` | `a && b` | `a && b` | `a and b` |
| logical OR | `a \|\| b` | `a \|\| b` | `a \|\| b` | `a \|\| b` | `a \|\| b` | `a or b` |
| logical NOT | `!a` | `!a` | `!a` | `!a` | `!a` | `not a` |
| short-circuits | yes | yes | yes | yes | yes | yes |
| `&&`/`\|\|` returns | `Boolean` | `Boolean` | `Boolean` | one of the operand values | same | one of the operand values |
| accepts non-`Boolean` in `if` | no | no | no | yes *(truthiness)* | yes | yes |

**The footgun.** In JavaScript and Python, `&&` / `and` and `\|\|` / `or` do *not* return a boolean — they return one of the operand values, chosen by truthiness. `"" \|\| "fallback"` in JavaScript returns `"fallback"`. `0 or "fallback"` in Python returns `"fallback"`. This is the canonical default-value idiom in dynamic languages — and the canonical source of bugs when `0` or `""` is a valid input.

### What counts as falsy?

| Language | Falsy values |
|---|---|
| Java / Scala / Kotlin | only `false`. Conditionals require a `Boolean`. |
| JavaScript / TypeScript | `false`, `0`, `-0`, `0n`, `""`, `null`, `undefined`, `NaN` |
| Python | `False`, `None`, `0`, `0.0`, `""`, `[]`, `{}`, `set()`, `()`, anything whose `__bool__` returns `False` or `__len__` returns `0` |

Everything else is truthy. Python's collection-emptiness rule (`if not my_list:`) is the idiom that surprises Java/Kotlin programmers the most.

### Statement vs expression

*Expression-oriented* languages (Scala, Kotlin) let `if`, `try`, and pattern-match constructs produce a value. *Statement-oriented* languages (Java pre-14, JS, Python) require ternary or workarounds.

| Construct | Java | Scala | Kotlin | JavaScript | TypeScript | Python |
|---|---|---|---|---|---|---|
| `if` produces a value | switch-expr *(14+)* | yes | yes | — *(use ternary)* | — | — *(use `a if c else b`)* |
| `try` produces a value | — | yes | yes | — | — | — |
| `switch` / `match` / `when` produces a value | switch-expr *(14+)* | yes *(`match`)* | yes *(`when`)* | — | — | yes *(`match`, 3.10+)* |
| block produces last value | — | yes *(`{ …; result }`)* | yes *(`run { … }`)* | — | — | — |
| function literal | lambda | lambda *(plus `def`)* | lambda | function expression | same | `lambda` *(single expression only)* |

## Notes — when a cell isn't enough

**Integer division on the JVM.** `1 / 2` is `0` in Java, Scala, Kotlin when both operands are ints. To get a float, at least one operand must be a float (`1 / 2.0` or `(double) 1 / 2`). JavaScript/TypeScript don't have this trap — all numbers are doubles. Python avoided it by making `/` always return float since 3.x, with `//` for floor division. Where it bites most: `(a + b) / 2` averages of odd numbers truncate silently.

**Dynamic-language `&&` / `\|\|` return values, not booleans.** `x \|\| default` is the classic default-value idiom in JavaScript and Python, but breaks the moment `x` is `0` or `""`. ES2020 added `??` precisely because `\|\|` had too many false positives. Modern JavaScript: use `??` for nullish fallback, `\|\|` only when you genuinely want truthy fallback. Python is stuck with `or` for both — the explicit form is `x if x is not None else default`.

**Elvis vs null-coalescing.** Kotlin's `?:` (Elvis) and JS/TS's `??` are the same idea — left if non-null, right otherwise. Both trigger only on null/undefined, not on `0` or `""`. Scala uses `Option.getOrElse`. Python convention is `x if x is not None else default` or `x or default` (the latter shares `\|\|`'s truthiness trap).

**Kotlin `!!` vs TypeScript `!`.** Kotlin's `!!` is a *runtime* check — it throws `NullPointerException` immediately if the value is null. TypeScript's `!` is a *static-only* non-null assertion — it tells the type-checker "trust me" and emits no runtime check. If TypeScript's `!` is wrong, the failure surfaces wherever you next use the value. Use both sparingly.

**No `++` / `--` in Scala, Kotlin, Python.** Scala and Kotlin omit increment/decrement operators deliberately — they encourage `i += 1` and prevent the prefix-vs-postfix confusion. Python omits them because of expression-statement separation: `i++` would be ambiguous to parse in a statement-oriented language.

**Unsigned right shift `>>>`.** Java, Scala, Kotlin (as `ushr`), JavaScript all have it. Python does not — Python ints are arbitrary-precision with no fixed bit width, so "unsigned" has no meaning. When porting bit-twiddling code to Python, mask explicitly: `(x >> n) & 0xFFFFFFFF`.

**Kotlin bitwise as infix functions, not operators.** Kotlin avoids overloading `&`, `\|`, `^` because the conventional usage (`&&`, `\|\|`) is logical. Instead Kotlin uses `and`, `or`, `xor`, `shl`, `shr`, `ushr`, `inv()`. Reads naturally — `mask and bits` instead of `mask & bits`.

**Chained comparison in Python.** `1 < x < 10` parses as `(1 < x) and (x < 10)`. No other language here does this — they all parse `a < b < c` as `(a < b) < c`, which then compares a boolean with `c`. Python's chained comparison is one of the small idioms it gets uniquely right.

**Walrus `:=` (Python 3.8+).** Lets you assign as part of an expression. Mostly useful in comprehensions and `while` loops: `while (line := f.readline()):`. Closest Python gets to a value-producing assignment. JavaScript's comma-operator `(x = …, x)` is the legacy equivalent — both are intentionally awkward to discourage overuse.

**Expression-orientation is a design philosophy, not a feature.** Scala and Kotlin treat almost everything as an expression — `if`, `try`, `match`/`when`, even blocks. Java is statement-oriented by tradition, slowly adding expression forms (switch expression in 14+, instanceof pattern, etc.). JavaScript and TypeScript allow ternary and arrow-function expressions but most control flow is statement-only. Python's expression set is small and intentional: `a if c else b`, `lambda`, comprehensions, walrus. This shapes how you structure code — expression-oriented languages reward composing rather than mutating.

**Operator overloading.** Scala allows defining nearly any symbol as a method (`def +(other: …)`). Kotlin allows it but only for a fixed set (`operator fun plus`, `times`, `minus`, etc.). Python uses dunder methods (`__add__`, `__mul__`, `__eq__`). Java does not allow operator overloading at all — the lone exception is `+` for `String` concatenation, which is hard-coded in the compiler. JavaScript/TypeScript have no operator overloading.
