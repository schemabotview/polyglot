The standard sequence, mapping, and set types — and the string type that every language implements differently underneath. Literal syntax, mutability defaults, indexing, slicing, and the operations every program reaches for.

## Sequences

Every language has at least two sequence types — the low-level array and the higher-level list / vector. The loudest divergence is **mutability defaults**: Java mutable, Scala immutable-by-default, JavaScript mutable, Python mutable (but `tuple` is immutable). Kotlin splits on the type — `List` (read-only) vs `MutableList`.

| Sub-feature | Java | Scala | Kotlin | JavaScript | TypeScript | Python |
|---|---|---|---|---|---|---|
| array literal | `new int[]{1,2,3}` | `Array(1, 2, 3)` | `arrayOf(1, 2, 3)` / `intArrayOf(...)` | `[1, 2, 3]` | same | — *(use `list`)* |
| default sequence | `ArrayList<T>` | `List` *(immutable, linked)* | `List<T>` *(read-only view)* | `Array` | same | `list` |
| immutable / read-only | `List.of(1, 2, 3)` *(9+)* | `List(1, 2, 3)` *(default)* | `listOf(1, 2, 3)` | `Object.freeze([...])` | `readonly T[]` | `tuple` `(1, 2, 3)` |
| mutable | `new ArrayList<>(...)` | `ArrayBuffer(...)` | `mutableListOf(...)` | `[1, 2, 3]` *(default)* | same | `list` *(default)* |
| index | `xs[0]` *(array)* / `xs.get(0)` *(list)* | `xs(0)` | `xs[0]` | `xs[0]` | same | `xs[0]` |
| negative index | — | — | — | — *(use `xs.at(-1)`)* | same | `xs[-1]` |
| length | `xs.length` *(array)* / `.size()` *(list)* | `xs.length` *(array)* / `.size` *(list)* | `xs.size` | `xs.length` | same | `len(xs)` |
| slice | `Arrays.copyOfRange(xs, i, j)` | `xs.slice(i, j)` | `xs.subList(i, j)` | `xs.slice(i, j)` | same | `xs[i:j]` |
| append | `xs.add(x)` *(list)* | `xs :+ x` / `xs += x` *(buffer)* | `xs + x` / `xs.add(x)` *(mutable)* | `xs.push(x)` | same | `xs.append(x)` |
| concat | `Stream.concat` / `addAll` | `xs ++ ys` | `xs + ys` | `[...xs, ...ys]` | same | `xs + ys` |
| contains | `xs.contains(x)` | `xs.contains(x)` | `x in xs` | `xs.includes(x)` | same | `x in xs` |
| empty literal | `List.of()` / `new ArrayList<>()` | `Nil` / `List.empty` | `emptyList()` | `[]` | same | `[]` |

Python is the only language with negative indexing baked in — `xs[-1]` is the last element. Everywhere else, you write `xs[xs.length - 1]` or call `last()`. JavaScript added `Array.prototype.at(-1)` in ES2022 to fix the asymmetry; the older `xs[-1]` returns `undefined` because arrays-are-objects and `-1` is just a property name that does not exist.

## Maps & Sets

### Map / dict / object

|---|---|---|---|---|---|---|

### Set

| Sub-feature | Java | Scala | Kotlin | JavaScript | TypeScript | Python |
|---|---|---|---|---|---|---|
| literal | `Set.of(1, 2, 3)` *(9+)* | `Set(1, 2, 3)` | `setOf(1, 2, 3)` | `new Set([1, 2, 3])` | same | `{1, 2, 3}` |
| empty literal | `Set.of()` | `Set.empty[T]` | `emptySet()` | `new Set()` | same | `set()` *(not `{}` — that's dict)* |
| membership | `s.contains(x)` | `s.contains(x)` | `x in s` | `s.has(x)` | same | `x in s` |
| union | manual / Guava | `a ++ b` / `a union b` | `a + b` / `a.union(b)` | manual `new Set([...a, ...b])` | same | `a | b` |

**Empty-literal trap.** In Python, `{}` is an empty *dict*, not a set — write `set()` for an empty set. In JavaScript, `{}` is an empty *object*; there is no set literal at all, you write `new Set()`. Memorize these — they are the most common one-character footguns in this whole topic.

## Strings — Type & Access

The string type is the most universally implemented type across these six — and the most divergent underneath. Internal encoding (UTF-16 vs flexible code-point storage), what counts as a *character*, and length semantics all differ.

| Sub-feature | Java | Scala | Kotlin | JavaScript | TypeScript | Python |
|---|---|---|---|---|---|---|
| type | `String` | `String` *(JVM)* | `String` *(JVM)* | `string` *(primitive)* | same | `str` |
| mutability | immutable | immutable | immutable | immutable | immutable | immutable |
| internal encoding | UTF-16 *(`char` = 16 bits)* | UTF-16 | UTF-16 | UTF-16 | same | flexible *(Latin-1 / UCS-2 / UCS-4 — PEP 393)* |
| character type | `char` *(16-bit)* | `Char` | `Char` | — *(no char — single-char string)* | same | — *(no char — single-char `str`)* |
| length | `s.length()` *(UTF-16 units)* | `s.length` *(UTF-16 units)* | `s.length` *(UTF-16 units)* | `s.length` *(UTF-16 units)* | same | `len(s)` *(code points)* |
| code-point count | `s.codePointCount(0, s.length())` | `s.codePointCount(...)` | `s.codePointCount(...)` | `[...s].length` *(spread iterates code points)* | same | `len(s)` |
| index | `s.charAt(i)` | `s(i)` | `s[i]` | `s[i]` *(or `.charAt(i)`)* | same | `s[i]` |
| negative index | — | — | — | — *(use `.at(-1)`)* | same | `s[-1]` |
| slice / substring | `s.substring(i, j)` | `s.substring(i, j)` / `s.slice(i, j)` | `s.substring(i, j)` / `s.substring(i..j)` | `s.slice(i, j)` / `s.substring(i, j)` | same | `s[i:j]` |
| concat | `s + t` *(or `.concat`)* | `s + t` | `s + t` | `s + t` | same | `s + t` |
| equality | `s.equals(t)` *(`==` is reference!)* | `==` *(structural via `equals`)* | `==` *(structural)* / `===` *(reference)* | `===` *(value)* | same | `==` *(value)* |

The single most-bitten footgun in this whole topic is **Java's `==` on strings**. `"hello" == "hello"` looks like it should work — and often does, because the JVM string-literal pool interns both literals to the same instance. But it is comparing references, and the moment a string is built at runtime the equality breaks. Use `.equals()`. Scala fixed this by routing `==` through `.equals()` for any non-null reference type; Kotlin's `==` is structural with `===` reserved for reference identity.

The second-biggest footgun is **UTF-16 length** in Java, Scala, Kotlin, and JavaScript. A string containing a single emoji outside the basic multilingual plane reports `length == 2` because it occupies a UTF-16 surrogate pair. Python's `len` counts code points, which is closer to *characters* but still not grapheme clusters — a flag emoji is two code points, a family emoji is many. There is no universally correct answer; only the answer that matches the question you are actually asking.

## Strings — Building & Formatting

| Sub-feature | Java | Scala | Kotlin | JavaScript | TypeScript | Python |
|---|---|---|---|---|---|---|
| interpolation | — *(use `String.format`)* | `s"value is $x"` / `s"${expr}"` | `"value is $x"` / `"${expr}"` | `` `value is ${x}` `` *(template)* | same | f-string `f"value is {x}"` |
| typed format | `String.format("%05d", x)` | `f"$x%05d"` *(typed at compile time)* | `"%05d".format(x)` | manual | same | `f"{x:05d}"` |
| multi-line | text block `"""..."""` *(13+)* | `"""..."""` | `"""..."""` *(raw)* | `` `...` `` *(template)* | same | `"""..."""` |
| raw string | — | `raw"..."` *(prefix)* | `"""..."""` *(no escape)* | `String.raw` `` `...` `` | same | `r"..."` |
| join collection | `String.join(",", xs)` | `xs.mkString(",")` | `xs.joinToString(",")` | `xs.join(",")` | same | `",".join(xs)` |
| split *(arg type)* | regex | regex | regex *(or `Char` overload for literal)* | string literal or regex | same | string literal *(use `re.split` for regex)* |
| repeat | `s.repeat(n)` *(11+)* | `s * n` | `s.repeat(n)` | `s.repeat(n)` | same | `s * n` |
| build incrementally | `StringBuilder` | `StringBuilder` | `buildString { ... }` | `arr.join('')` *(or template)* | same | `''.join(parts)` |
| trim | `s.strip()` *(Unicode aware, 11+)* | `s.trim` | `s.trim()` | `s.trim()` | same | `s.strip()` |
| case change | `s.toUpperCase()` | `s.toUpperCase` | `s.uppercase()` | `s.toUpperCase()` | same | `s.upper()` |

**Build incrementally — the universal *do not concatenate in a loop* warning.** Strings are immutable in every one of these six, so `s = s + chunk` inside a loop is O(n²) — each `+` allocates a new buffer and copies both sides. The fix is to collect into a list and join once at the end, or use a mutable builder. Python's `''.join(parts)` is the idiomatic form. Java's `StringBuilder` is what `javac` turns `+` into when the chain is straight-line — but the moment a loop is involved, the optimization breaks down and you must use `StringBuilder` explicitly.

**Splitting on a regex by default — Java / Scala / Kotlin.** `s.split(".")` in those three matches *every character* because `.` is a regex metacharacter. Splitting on `|` matches between every character. The fix is `Pattern.quote` in Java, or pass a `Char` (single quotes) instead of a `String` in Kotlin. Python's `s.split(".")` is literal by default; use `re.split` for regex. JavaScript accepts either — a string is literal, a regex object is regex.

## Notes — when a cell isn't enough

**Scala collections — immutable by default, mutable on opt-in.** The `scala.collection.immutable` package is the default — plain `List` resolves to `immutable.List`. Mutating operations on these return *new* collections; the original is untouched. The `scala.collection.mutable` package mirrors the same names — `mutable.Map`, `mutable.ArrayBuffer`, `mutable.Set` — for in-place mutation. Idiomatic Scala defaults to immutable and only reaches for mutable inside a small scope. This is the largest single mental-model shift coming from Java.

**Kotlin's `List` is read-only, not immutable.** `List<T>` in Kotlin is the *read-only* interface — you cannot call `.add` on it. But the underlying instance may be a `MutableList` that someone else is mutating. So `List<T>` gives a guarantee about *your* call site, not about the value. For true immutability, copy on input or use `kotlinx.collections.immutable`. Java's `List.of(...)` (since 9) returns a genuinely immutable list — calling `.add` throws `UnsupportedOperationException`.

**Java boxing on the collection boundary.** Java collections are generic and generics are erased to `Object`. So `List<Integer>` boxes every `int` into an `Integer` on insert and unboxes on read. Real performance hit on hot paths — workarounds are `int[]`, `IntStream`, or third-party primitive-specialized collections like Eclipse Collections or fastutil. The other five do not have this problem because they either have no primitives (JavaScript, Python) or their primitive arrays bypass generics (Kotlin's `IntArray`, Scala's `Array[Int]`).

**JavaScript objects vs Maps.** Pre-ES6, `Object` was the only key-value type. Keys had to be strings (numbers were coerced). Iteration order was implementation-defined. ES6 added the proper `Map` type — any-typed keys, insertion-ordered, has a `size` property. Use `Map` when keys are not statically known strings. Use `Object` (or `Record<K, V>` in TypeScript) when you have a fixed schema. Common bug: using an object as a hash map and being surprised by `__proto__`, `constructor`, or `hasOwnProperty` colliding with real keys.

**Python dict ordering — guaranteed since 3.7.** Python's `dict` preserves insertion order as of 3.7 (CPython implementation since 3.6, language guarantee since 3.7). Before that, you needed `collections.OrderedDict`. This is a permanent guarantee — not an implementation detail. JavaScript's `Map` has the same insertion-order guarantee since ES6. Java's `LinkedHashMap` opts in; plain `HashMap` does not.

**Empty-literal collisions.** Python: `{}` is an empty *dict*; `set()` for an empty set. JavaScript: `{}` is an empty *object*; `new Set()` for an empty set, `new Map()` for an empty map. Kotlin: `mapOf()`, `setOf()`, `listOf()` are all unambiguous. Scala: `Map.empty[K, V]` and `Set.empty[T]` need type parameters because there is nothing to infer. Java: `Map.of()` / `Set.of()` / `List.of()` all work and are immutable.

**Java string `==` and the string pool.** `"hello" == "hello"` returns `true` because both literals point at the same interned instance in the JVM string pool. `"hel" + "lo" == "hello"` may also return `true` if the compiler folds the constant. But `new String("hello") == "hello"` is `false`. And once you read strings from a file or network, all bets are off — they are not pooled. **Always use `.equals()` for string equality in Java.** Scala's `==` routes through `.equals()`; Kotlin's `==` is structural with `===` for reference identity.

**UTF-16 vs UTF-8 vs code points vs grapheme clusters.** Java, Scala, Kotlin, and JavaScript all store strings as UTF-16 internally, so `.length` counts UTF-16 code units. A character outside the basic multilingual plane (most modern emoji, some CJK extensions) is a *surrogate pair* — two code units encoding one code point. So `"🎉".length === 2` in JavaScript. Python's `len` counts code points (PEP 393 stores strings in the most compact form that fits — Latin-1, UCS-2, or UCS-4 — but exposes them uniformly). Neither is *grapheme clusters* — a flag emoji is two regional-indicator code points; a family-of-four emoji is many code points joined by zero-width joiners. For grapheme-cluster correctness, use a Unicode-aware library: ICU on the JVM, `Intl.Segmenter` in modern JavaScript, the `regex` package in Python.

**String interpolation gotchas.** Scala's `s"..."` is the basic interpolator; `f"..."` adds typed format specifiers checked at compile time — `f"$x%05d"` will fail to compile if `x` is not numeric. Kotlin's `"$x"` interpolates a simple identifier; `"${expr}"` interpolates an expression — and works inside any string literal, no prefix. JavaScript template literals use **backticks**, not double quotes. Python f-strings (3.6+) allow expressions including method calls and slicing — `f"{name.upper()[:3]}"`. Java has no built-in interpolation — use `String.format` or text-block `.formatted()`. Java's template-string proposal exists but has bounced between preview and removal.

**Slicing semantics — half-open everywhere.** `xs[i:j]` and `xs.slice(i, j)` are half-open in all six: index `i` inclusive, index `j` exclusive. Python is the most flexible: `xs[i:j:k]` adds a step, `xs[::-1]` reverses, missing indices default to *to the end* or *from the start*. The functional-style `take(n)` and `drop(n)` exist in Scala, Kotlin, and via libraries elsewhere — often clearer than slice indices when the intent is *first N* or *skip first N*.

**Multi-line and raw strings.** Java text blocks since 13 — triple-quoted with smart leading-whitespace stripping based on the closing delimiter's indentation. Scala and Kotlin both use `"""..."""` for raw multi-line. Kotlin's `.trimIndent()` and `.trimMargin()` post-process the indentation. JavaScript template literals (backticks) preserve every whitespace character literally including indentation — surprising the first time inside an indented block. Python triple-quoted strings preserve whitespace too; `textwrap.dedent` is the cleanup tool.
